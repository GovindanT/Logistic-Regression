# 1. Import Necessary Libraries

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
```

# 2. Data Collection

```python
claimants_data = pd.read_csv(r"C:\Users\gthangar\OneDrive - Nokia\1830_Automation\ML\Task\Machine_Learning\03-Logistic_Regression\claimants.csv")
claimants_data
```

# 4. Data Preparation

```python
claimants_data.dropna(inplace=True)
```

```python
claimants_data.shape
```

```python
X = claimants_data.drop(labels=['CASENUM','ATTORNEY'], axis=1)
X
```

```python
y = claimants_data['ATTORNEY']
y
```

# 5. Model Building

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
```

```python
print("Training data :", len(X_train),len(y_train))
print("Testing data :", len(X_test),len(y_test))
```

```python
logistic_model = LogisticRegression()
```

# 6. Model Training

```python
logistic_model.fit(X_train,y_train)
```

# 7. Model Testing

### Training Data

```python
y_pred_train = logistic_model.predict(X_train)
y_pred_train
```

### Testing Data

```python
y_pred_test = logistic_model.predict(X_test)
y_pred_test
```

# 8. Model Evaluation

```python
from sklearn.metrics import classification_report,confusion_matrix,accuracy_score
```

```python
print(classification_report(y_true = y_train,y_pred = y_pred_train))
```

```python
confusion_matrix(y_true=y_train, y_pred=y_pred_train)
```

```python
print(classification_report(y_true = y_test,y_pred = y_pred_test))
```

```python
confusion_matrix(y_true=y_test, y_pred=y_pred_test)
```
